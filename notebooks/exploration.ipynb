```python
import sys
sys.path.insert(0, '..')
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Browse dataset structure
dataset_dir = '../data/leapGestRecog'
classes = sorted(os.listdir(dataset_dir))
print('Gesture classes:', classes)
```

```python
# Display one sample per class
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.flatten()

for idx, cls in enumerate(classes[:10]):
    cls_path = os.path.join(dataset_dir, cls)
    for root, _, files in os.walk(cls_path):
        for f in files:
            if f.lower().endswith('.png'):
                img = cv2.imread(os.path.join(root, f), cv2.IMREAD_GRAYSCALE)
                axes[idx].imshow(img, cmap='gray')
                axes[idx].set_title(cls)
                axes[idx].axis('off')
                break
        break

plt.tight_layout()
```
